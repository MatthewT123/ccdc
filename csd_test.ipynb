```python
from ccdc.conformer import MoleculeMinimiser, ConformerGenerator
from ccdc.io import MoleculeWriter
from ccdc import io

from utils.visualize import visualise_mol, visualise_conformers
```

```python
# Read from CSD
reader = io.EntryReader("CSD")
mol = reader.entry("DINXIJ").molecule
big = max(mol.components, key=lambda c: len(list(c.atoms)))
mol = big

# Find the minimized structure
minimiser = MoleculeMinimiser()
minimised_mol = minimiser.minimise(mol)

visualise_mol(mol, minimised_mol)
```

```python
# Generate some conformers
conformer_generator = ConformerGenerator()
conformers = conformer_generator.generate(mol)

with MoleculeWriter('conformers.mol2') as mol_writer:
    for c in conformers:
        mol_writer.write(c.molecule)
```

### Let's display some of the conformers we've generated!

```python
conformers = list(ConformerGenerator().generate(mol))  
visualise_conformers(conformers[:20], align=True)
```
